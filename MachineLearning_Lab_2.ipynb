```markdown
## Part I: Working with Dictionary

A data structure used to store data in key-value pairs.
```

````python
# Create a dictionary
student = {
    "name": "Ram",
    "age": 20,
    "faculty": "BCE"
}
display(f"Initial Dictionary: {student}")

d = {"a": 1, "b": 2}
display(f"New Dictionary d: {d}")
````

````python
# Access Value
display(f"Accessing 'name' from student: {student['name']}")

# Use get()
display(f"Using get() for 'age' from student: {student.get('age')}")
display(f"Using get() for non-existent key 'gender': {student.get('gender')}")
````

````python
# Add Item
student["grade"] = "A"
display(f"Dictionary after adding 'grade': {student}")
d['c'] = 3
display(f"Dictionary d after adding 'c': {d}")
````

````python
# Update Value
student["age"] = 21
display(f"Dictionary after updating 'age': {student}")
d["a"] = 10
display(f"Dictionary d after updating 'a': {d}")
````

````python
# Remove Item using del
if "faculty" in student:
    del student["faculty"]
display(f"Dictionary after deleting 'faculty': {student}")

if "b" in d:
    del d["b"]
display(f"Dictionary d after deleting 'b': {d}")
````

````python
# Remove using pop()
popped_grade = student.pop("grade")
display(f"Popped 'grade': {popped_grade}, Dictionary: {student}")

popped_a = d.pop("a")
display(f"Popped 'a': {popped_a}, Dictionary d: {d}")
````

````python
# Remove Last Item d.popitem()
if student: # Ensure dictionary is not empty
    popped_item = student.popitem()
    display(f"Popped last item: {popped_item}, Dictionary: {student}")
else:
    display("Student dictionary is empty, cannot popitem.")

if d: # Ensure dictionary is not empty
    popped_item_d = d.popitem()
    display(f"Popped last item from d: {popped_item_d}, Dictionary d: {d}")
else:
    display("Dictionary d is empty, cannot popitem.")
````

````python
# Copy Dictionary
original_d = {"x": 10, "y": 20}
new_d = original_d.copy()
display(f"Original dictionary for copy: {original_d}")
display(f"Copied dictionary new_d: {new_d}")

# Modify new_d to show it's a separate copy
new_d["z"] = 30
display(f"Modified new_d: {new_d}")
display(f"Original d remains unchanged: {original_d}")
````

````python
# Reinitialize student dictionary for further operations
student = {
    "name": "Ram",
    "age": 20,
    "faculty": "BCE",
    "grade": "A"
}
display(f"Reinitialized student dictionary: {student}")

# Get All Keys
display(f"All keys: {student.keys()}")

# Get All Values
display(f"All values: {student.values()}")

# Get All Items
display(f"All items: {student.items()}")
````

````python
# Check Key Exists
display(f"Is 'name' in student? {'name' in student}")
display(f"Is 'address' in student? {'address' in student}")
````

````python
# Get Length
display(f"Length of student dictionary: {len(student)}")
````

````python
# Loop Through Keys
display("Looping through keys:")
for k in student:
    display(k)

# Loop Through Values
display("Looping through values:")
for v in student.values():
    display(v)

# Loop Through Items
display("Looping through items:")
for k, v in student.items():
    display(f"{k}: {v}")
````

```markdown
## Part II: Working with Pandas DataFrame

A DataFrame in Python is a two-dimensional, tabular data structure that organizes data into labeled rows and columns. It is conceptually identical to an Excel spreadsheet or a SQL table, making it the primary tool for data manipulation and analysis in the Pandas Library.
```

````python
import pandas as pd

# Create a DataFrame containing student details
data = {
    'Name': ['Ram', 'Shyam', 'Hari', 'Sita', 'Gita'],
    'Age': [20, 22, 19, 21, 23],
    'Marks': [85, 78, 92, 65, 88]
}
df = pd.DataFrame(data)

display("Created DataFrame:")
display(df)
````

````python
# Display the first 5 rows of the DataFrame.
display("First 5 rows:")
display(df.head())
````

````python
# Display the last 5 rows of the DataFrame.
display("Last 5 rows:")
display(df.tail())
````

````python
# Find the total number of rows and columns in the DataFrame.
display("Shape (rows, columns):")
display(df.shape)
````

````python
# Display all column names in the DataFrame.
display("Column names:")
display(df.columns)
````

````python
# Display complete information about the DataFrame including data types and memory usage.
display("DataFrame Info:")
df.info()
````

````python
# Generate a statistical summary of numerical columns in the DataFrame.
display("Statistical Summary:")
display(df.describe())
````

````python
# Select and display the Name column from the DataFrame.
display("Name column:")
display(df['Name'])
````

````python
# Select and display multiple columns such as Name and Marks.
display("Name and Marks columns:")
display(df[['Name', 'Marks']])
````

````python
# Display the first row of the DataFrame using positional indexing.
display("First row using iloc:")
display(df.iloc[0])
````

````python
# Display the first row of the DataFrame using label-based indexing.
display("First row using loc:")
display(df.loc[0])
````

````python
# Access a specific cell value such as Marks of the first student.
display("Marks of the first student (Ram):")
display(df.loc[0, 'Marks'])
````

````python
# Filter and display students whose Marks are greater than 80.
display("Students with Marks > 80:")
display(df[df['Marks'] > 80])
````

````python
# Filter and display students whose Age is greater than 20 and Marks are greater than 75.
display("Students with Age > 20 AND Marks > 75:")
display(df[(df['Age'] > 20) & (df['Marks'] > 75)])
````

````python
# Add a new column called Grade to the DataFrame.
# Ensure the length of the list matches the number of rows
df['Grade'] = ['A', 'B', 'A', 'C', 'A']
display("DataFrame after adding 'Grade' column:")
display(df)
````

````python
# Increase all students’ Marks by 5.
df['Marks'] = df['Marks'] + 5
display("DataFrame after increasing Marks by 5:")
display(df)
````

````python
# Update the Marks of Ram to 95 using a condition.
df.loc[df['Name'] == 'Ram', 'Marks'] = 95
display("DataFrame after updating Ram's Marks to 95:")
display(df)
````

````python
# Remove the Age column from the DataFrame.
df_no_age = df.drop('Age', axis=1)
display("DataFrame after removing 'Age' column (original df is unchanged unless reassigned):")
display(df_no_age)
# Note: df.drop returns a new DataFrame; the original df remains unchanged unless reassigned
display("Original DataFrame (Age column still exists):")
display(df)
````

````python
# Remove the first row from the DataFrame.
df_no_first_row = df.drop(0, axis=0)
display("DataFrame after removing the first row (original df is unchanged unless reassigned):")
display(df_no_first_row)
````

````python
# Sort the DataFrame by Marks in ascending order.
display("DataFrame sorted by Marks (ascending):")
display(df.sort_values('Marks'))
````

````python
# Sort the DataFrame by Marks in descending order.
display("DataFrame sorted by Marks (descending):")
display(df.sort_values('Marks', ascending=False))
````

````python
# Concatenate two DataFrames vertically.
data2 = {
    'Name': ['Prem', 'Laxmi'],
    'Age': [24, 25],
    'Marks': [70, 90],
    'Grade': ['B', 'A']
}
df2 = pd.DataFrame(data2)

display("Second DataFrame to concatenate:")
display(df2)

df_concatenated = pd.concat([df, df2])
display("Concatenated DataFrame:")
display(df_concatenated)
````

````python
# Read data from a CSV file into a DataFrame.
# First, let's save our current DataFrame to a CSV to demonstrate reading it back.
df.to_csv('students.csv', index=False)
display("DataFrame saved to students.csv. Now reading it back:")

df_from_csv = pd.read_csv('students.csv')
display(df_from_csv)
````

````python
# Write the DataFrame into a CSV file.
# This was already demonstrated in the previous step, but explicitly here:
df_concatenated.to_csv('all_students.csv', index=False)
display("Concatenated DataFrame saved to all_students.csv.")
# You can check the file in the Colab file browser.
````

````python
# Read data from an Excel file into a DataFrame.
# First, let's save our current DataFrame to an Excel to demonstrate reading it back.
# You might need to install openpyxl if not already present:
# !pip install openpyxl
df.to_excel('students.xlsx', index=False)
display("DataFrame saved to students.xlsx. Now reading it back:")

df_from_excel = pd.read_excel('students.xlsx')
display(df_from_excel)
````

````python
# Write the DataFrame into an Excel file.
# This was already demonstrated in the previous step, but explicitly here:
df_concatenated.to_excel('all_students.xlsx', index=False)
display("Concatenated DataFrame saved to all_students.xlsx.")
# You can check the file in the Colab file browser.
````
