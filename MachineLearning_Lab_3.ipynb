## Part E: Outlier Detection

### Task 15: Draw boxplot.

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x=df['Fare'])
plt.title('Boxplot of Fare')
plt.show()
```

### Task 16: Calculate Inter Quartile Range.

```python
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1 (25th percentile): {Q1:.2f}")
print(f"Q3 (75th percentile): {Q3:.2f}")
print(f"IQR (Interquartile Range): {IQR:.2f}")
print(f"Lower bound for outliers: {lower_bound:.2f}")
print(f"Upper bound for outliers: {upper_bound:.2f}")
```

### Task 17: Find outliers.

```python
outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]
print(f"Number of outliers in 'Fare' column: {len(outliers)}")
print('Outliers in Fare column:')
display(outliers[['Fare']])
```

### 4. Why should missing values be handled?

Missing values need to be handled because:
1. **Many machine learning algorithms cannot handle missing values** directly and will throw an error or behave unexpectedly.
2. **They can lead to biased or inefficient models.** Leaving missing values can lead to inaccurate conclusions if the missingness is not random.
3. **They reduce the statistical power of the model.** Datapoints with missing values might be completely removed if not handled, leading to a loss of valuable information.
4. **They can distort analysis.** For example, calculating an average on a column with many missing values might not be representative.

Handling missing values ensures the dataset is clean, complete, and suitable for model training, leading to more robust and accurate predictions.

### 5. Why use median instead of mean for Age?

The median is often preferred over the mean for imputing 'Age' (or other skewed numerical features) because:
1. **Robustness to Outliers:** Age distribution can be skewed, and there might be outliers (e.g., very young children or very old individuals). The mean is sensitive to outliers and extreme values, which can pull it significantly in one direction, making it a less representative central tendency. The median, being the middle value, is less affected by these extreme values.
2. **Maintaining Distribution Shape:** Using the median helps preserve the overall shape of the age distribution better than the mean, especially in skewed distributions.

Since age distributions are often slightly skewed (e.g., more young people than very old people, or a peak in middle age), the median provides a more robust and representative central value for imputation.

### 7. Why can't machine learning algorithms directly use text values?

Machine learning algorithms are fundamentally mathematical and operate on numerical data. They cannot directly interpret or process raw text values for several reasons:
1. **Mathematical Operations:** Algorithms perform calculations (e.g., distance, multiplication, summation) on data. Textual labels do not have inherent mathematical properties that can be used in these operations.
2. **Interpretation of Relationships:** Algorithms need to understand relationships between data points. Text labels like 'male' and 'female' or 'Southampton' and 'Cherbourg' don't inherently convey a quantitative difference or order that the algorithm can use.
3. **Memory and Efficiency:** Storing and processing text data is generally less efficient than numerical data.

Therefore, categorical text data must be converted into a numerical format (e.g., through Label Encoding or One-Hot Encoding) before being fed into most machine learning models.

### 8. What is the difference between Label Encoding and One-Hot Encoding?

**Label Encoding:**
- **Method:** Assigns a unique integer to each category in a feature. For example, ['red', 'green', 'blue'] might become [0, 1, 2].
- **Use Case:** Best suited for **ordinal categorical variables** where there is a natural order or ranking among the categories (e.g., 'low', 'medium', 'high'). It's also often used for **binary categorical variables** (like 'male'/'female') as it doesn't introduce an artificial order issue.
- **Disadvantage:** For nominal categorical variables (where no inherent order exists, like 'Embarked' locations), label encoding introduces an artificial ordinal relationship (e.g., 2 > 1 > 0). This can mislead models into thinking that higher numerical values imply higher importance or a greater magnitude, which is incorrect and can negatively impact performance.

**One-Hot Encoding:**
- **Method:** Creates new binary (0 or 1) columns for each category in a feature. If a category is present, the column for that category gets a 1, and others get 0. For example, ['red', 'green', 'blue'] might become: `red_col` (1,0,0), `green_col` (0,1,0), `blue_col` (0,0,1).
- **Use Case:** Best suited for **nominal categorical variables** where there is no inherent order between categories (e.g., 'Embarked' ports, 'color'). It avoids the problem of introducing artificial ordinality.
- **Disadvantage:** Can lead to a high-dimensional dataset if there are many categories, potentially causing a 'curse of dimensionality'. It also introduces multicollinearity if all new columns are kept (often `drop_first=True` is used to mitigate this).

### 9. Why is scaling necessary?

Feature scaling is a crucial preprocessing step for several reasons:
1. **Distance-Based Algorithms:** Many machine learning algorithms (e.g., K-Nearest Neighbors, Support Vector Machines, K-Means clustering) calculate distances between data points. If features have vastly different scales, features with larger values will dominate the distance calculation, making the smaller-scaled features almost irrelevant. Scaling ensures all features contribute equally to the distance.
2. **Gradient Descent Optimization:** Algorithms that use gradient descent (e.g., neural networks, logistic regression, linear regression) converge much faster when features are scaled. Unscaled features can lead to an elongated and winding cost function, making it harder and slower for the optimizer to find the minimum.
3. **Regularization Techniques:** Regularization methods (L1, L2) penalize large coefficients. If features are not scaled, features with larger ranges might get larger coefficients by chance, and the regularization penalty would disproportionately affect them.
4. **Interpretability:** Scaling can sometimes make coefficients more interpretable, as they relate to changes in standard deviations rather than raw units.

### 10. Which algorithms require scaling?

Algorithms that are particularly sensitive to feature scaling include:

*   **Distance-based algorithms:**
    *   K-Nearest Neighbors (KNN)
    *   Support Vector Machines (SVM) with radial basis function (RBF) or polynomial kernels
    *   K-Means Clustering
*   **Algorithms using gradient descent:**
    *   Linear Regression
    *   Logistic Regression
    *   Neural Networks (Deep Learning models)
*   **Algorithms that compute covariance or correlation:**
    *   Principal Component Analysis (PCA)
    *   Linear Discriminant Analysis (LDA)
*   **Regularized models:**
    *   Lasso and Ridge Regression (L1 and L2 regularization)

Algorithms that are generally **not sensitive** to feature scaling include tree-based models like Decision Trees, Random Forests, and Gradient Boosting Machines, as they make decisions based on feature thresholds rather than distances or magnitudes.

### 11. What is an outlier?

An **outlier** is a data point that significantly differs from other observations. It lies an abnormal distance from other values in a random sample from a population. Outliers can occur due to:
1.  **Measurement errors:** Mistakes during data collection.
2.  **Experimental errors:** Problems during an experiment.
3.  **Data entry errors:** Typos or incorrect data input.
4.  **Natural variation:** Genuine extreme values in the data that are rare but valid.
5.  **Intentional errors:** Data manipulation or fraudulent entries.

Outliers can be univariate (extreme on a single variable) or multivariate (extreme on a combination of variables). They are typically detected using statistical methods (like IQR rule, Z-score, or Isolation Forest) or visualization techniques (like boxplots).

### 12. Why are outliers problematic?

Outliers can be problematic for several reasons:
1.  **Distortion of Statistics:** Outliers can significantly skew descriptive statistics like the mean and standard deviation, making them unrepresentative of the majority of the data. The median, however, is more robust to outliers.
2.  **Impact on Machine Learning Models:**
    *   **Sensitivity:** Many machine learning algorithms (especially linear models, distance-based algorithms like K-Nearest Neighbors and SVMs, and neural networks) are highly sensitive to outliers. Outliers can lead to models that perform poorly on unseen data because they might learn to fit the extreme values rather than the general pattern.
    *   **Incorrect Decision Boundaries:** In classification tasks, outliers can pull decision boundaries away from the optimal position.
    *   **Overfitting:** Models might overfit to the noise introduced by outliers.
3.  **Violation of Assumptions:** Many statistical tests and models assume that data follows a certain distribution (e.g., normal distribution). Outliers can violate these assumptions, leading to incorrect inferences or less reliable models.
4.  **Misleading Visualizations:** Outliers can distort the scale of plots (e.g., histograms, scatter plots, box plots), making it difficult to visualize the distribution and relationships among the bulk of the data.
